```python
# Import Pandas
import pandas as pd

# Read Excel Files in the DataFrame
df = pd.read_excel("US_RETAIL_DATA.xlsx")

# Print DataFrame
print(df)
```

```python
# Variance for Each Columns (Columns - Vertically - axis=0 - default)
# df.var(numeric_only=True)

df.var(numeric_only=True)
```

```python
# Variance for Each Rows (Rows - Horizontally - axis=1)
# df.var(numeric_only=True,axis=1)

df.var(numeric_only=True,axis=1)
```

```python
# Population Variance - ddof=0
# ddof - Delta Degrees of Freedom
# df.var(ddof=0)

df.var(numeric_only=True,ddof=0)
```

```python
# Sample Variance - ddof=1 (default)
# ddof - Delta Degrees of Freedom
# df.var(ddof=1)

df.var(numeric_only=True,ddof=1)
```

```python
# Combination of Mean & Variance
# Accepts "Numerical Data" only
df.agg(['mean','var'])
```
