```python
# Import Pandas
import pandas as pd

# Read Excel Files in the DataFrame
df = pd.read_excel("US_RETAIL_DATA.xlsx")

# Print DataFrame
print(df)
```

```python
# Standard Deviation for Each Columns (Columns - Vertically - axis=0 - default)
# df.std(numeric_only=True)

df.std(numeric_only=True)
```

```python
# Standard Deviation for Each Rows (Rows - Horizontally - axis=1)
# df.std(numeric_only=True,axis=1)

df.std(numeric_only=True,axis=1)
```

```python
# Population Standard Deviation - ddof=0
# ddof - Delta Degrees of Freedom
# df.std(ddof=0)

df.std(numeric_only=True,ddof=0)
```

```python
# Sample Standard Deviation - ddof=1 (default)
# ddof - Delta Degrees of Freedom
# df.std(ddof=1)

df.std(numeric_only=True,ddof=1)
```

```python
# Combination of Mean & Standard Deviation
# Accepts "Numerical Data" only
df.agg(['mean','std'])
```
