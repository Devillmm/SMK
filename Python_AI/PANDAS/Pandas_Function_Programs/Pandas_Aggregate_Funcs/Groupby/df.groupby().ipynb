```python
# Import Pandas
import pandas as pd

# Read Excel Files in the DataFrame
df = pd.read_excel("US_RETAIL_DATA.xlsx")

# Print DataFrame
print(df)
```

```python
# Groupby by Columns
# df.groupby("Column_name") --> Object Output

Grp = df.groupby("Category")
print(Grp)
```

```python
# Groupby using Aggregate Functions
# group.Aggregate_Func()
print(Grp.sum(numeric_only=True))
```

```python
# Groupby using Aggregate Functions
print(Grp.mean(numeric_only=True))
```

```python
# Groupby using Aggregate Functions for Specific Columns
# group["Column_name"].Aggregate_Func()
print(Grp["Sales"].sum())
print()
print(Grp["Profit"].mean())
```
