```python
# Import Pandas
import pandas as pd

# Read .CSV File and Print the Records using "Default Index"
df = pd.read_csv("customers.csv")
print(df)
```

```python
# Read the .CSV File and Print the Records using "index_col"
# index_col --> DataFrame Records (any Column) act as an "Index"

df_col = pd.read_csv("customers.csv",index_col = "Customer Id")
print(df_col)
```

```python
# Print All Records in a Single Sheet

print(df_col.to_string())
```

```python
# Print any "Specific Rows" using "Customer Id" as Index Value
# print(df.loc["Unique_Id"]) - 'Index Location'

print(df_col.loc["BE91A0bdcA49Bbc"])
```

```python
# Print any "Specific Column Values" using "Customer Id as Index Location"
# print(df.loc["Unique_Id",['column_1','column_2',...]])

print(df_col.loc["BF6a1f9bd1bf8DE",["Last Name","City","Country"]])
```

```python
# Print Range-wise / Slicing some Rows & Columns
# print(df.loc["Unique_Id1":"Unique_Id6",['column_1','column_2',....]])

print(df_col.loc["216E205d6eBb815":"0F60FF3DdCd7aB0",["Last Name","City","Country"]])
```

```python
# Slicing using iloc - Integer Location
# 0:5 - Print Index 0 to 4 Rows (Exclusive)
# 1:3 - Print Columns 1 to 3 Columns (Inclusive)
print(df.iloc[0:5,1:3])
```

```python
# Slicing with Step Value using iloc - Integer Location
# Step Value-2 --> 0:5:2 = (0,2,4)
print(df.iloc[0:5:2,1:7])
```

```python
# Slicing with Step Value using iloc - Integer Location
# Both has Step Value-2 --> 1:7:2 = (0,2,4)
print(df.iloc[0:5:2,1:7:2])
```
