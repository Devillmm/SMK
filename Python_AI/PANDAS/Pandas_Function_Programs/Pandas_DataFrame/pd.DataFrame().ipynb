```python
# Import Pandas
import pandas as pd
```

```python
## Pandas DataFrame - Multi-Dimensional Array (Rows & Columns)
## Default Index - pd.DataFrame(ndArray)

# Make a Dictionary
Employees = {
    "Name":['Mani','Steve','Tony','Chris','Stark','Tom','Tobby'],
    "place":[100,110,120,130,135,140,146],
    "Salary":[10000,14000,15000,20000,35000,43000,25000]
}

# Print DataFrame with Default Index [df]
df = pd.DataFrame(Employees)
print("Employee Table with Default Index:\n",df)
```

```python
# Print DataFrame with Fixed Index
# pd.DataFrame(ndArray,index=[])

# Make a Dictionary
Employees = {
    "Name":['Mani','Steve','Tony','Chris','Stark','Tom','Tobby'],
    "place":[100,110,120,130,135,140,146],
    "Salary":[10000,14000,15000,20000,35000,43000,25000]
}

df = pd.DataFrame(Employees,index=['E1','E2','E3','E4','E5','E6','E7'])
print("Employees Table with Fixed Index:\n",df)
```

```python
# To Access 'Single Row' in DataFrame using "Index Location"
# df.loc['index_value'] --> "Index Location"

# Make a Dictionary
Employees = {
    "Name":['Mani','Steve','Tony','Chris','Stark','Tom','Tobby'],
    "place":[100,110,120,130,135,140,146],
    "Salary":[10000,14000,15000,20000,35000,43000,25000]
}

# Index Location
df = pd.DataFrame(Employees,index=['E1','E2','E3','E4','E5','E6','E7'])
print("Index Location:")
print(df.loc['E6'])
```

```python
# To Access 'Single Row' in DataFrame using "Integer Location"
# df.iloc['value'] --> "Integer Location"

# Make a Dictionary
Employees = {
    "Name":['Mani','Steve','Tony','Chris','Stark','Tom','Tobby'],
    "place":[100,110,120,130,135,140,146],
    "Salary":[10000,14000,15000,20000,35000,43000,25000]
}

# Integer Location
df = pd.DataFrame(Employees)
print("Integer Location:")
print(df.iloc[4])
```

```python
# Add "New Column" in the DataFrame
# df["New_Column_name"] = ['Value1','Value2','Value3',....]

# Make a Dictionary
Employees = {
    "Name":['Mani','Steve','Tony','Chris','Stark','Tom','Tobby'],
    "place":[100,110,120,130,135,140,146],
    "Salary":[10000,14000,15000,20000,35000,43000,25000]
}

# Create a DataFrame
df = pd.DataFrame(Employees)

# Add 'New Column' in the Existing DataFrame
df["Desigination"] = ["Data Engineer","Data Scientist","ML Developer","AI Developer","BI Engineer","SQL Techi","PowerBI Developer"]
print(df)
```

```python
# Create "Multiple Rows" in the DataFrame
# df_name = pd.DataFrame([{},{},...],index=[])

df_name = pd.DataFrame([{'Std_Name':"Jerry",'Age':15,'Course':"Data Analytics",'Marks':85},
             {'Std_Name':"Nobita",'Age':14,'Course':"Data visualization",'Marks':89},
             {'Std_Name':"Doremon",'Age':16,'Course':"Web Development",'Marks':90}],
             index = ['S1','S2','S3']
             )
print("Multiple Rows in the DataFrame:")
print(df_name)
```

```python
# Add "New Row" in the DataFrame & Concatenate the 'Two DataFrames'
# New Row --> df = pd.DataFrame([{}],index=[])
# Concatenate Two DataFrames --> df_name = pd.concat([df as list])

# Add "New Row" in the DataFrame
df = pd.DataFrame([{'Std_Name':"Martin",'Age':17,'Course':"Data Analytics",'Marks':87}],index=['S4'])
print("New Row Creation:")
print(df)
print()

# Previous DataFrame [df_name]
df_name = pd.DataFrame([{'Std_Name':"Jerry",'Age':15,'Course':"Data Analytics",'Marks':85},
             {'Std_Name':"Nobita",'Age':14,'Course':"Data visualization",'Marks':89},
             {'Std_Name':"Doremon",'Age':16,'Course':"Web Development",'Marks':90}],
             index = ['S1','S2','S3']
             )

# Concatenate Two DataFrames
df_name = pd.concat([df_name,df])
print("New DataFrame after Concatenation:")
print(df_name)
```
