```python
# Import Pandas
import pandas as pd
```

```python
# Pandas Series - Single Column 1D Array
# Print Series with Index --> "pd.Series(array,index=[])""
roll_no = [12,13,14,15]
students = pd.Series(roll_no,index=["S1","S2","S3","S4"])
print(students)
```

```python
# Print Series with Default Index - "pd.Series(array)"
age = [20,23,25,27]
series = pd.Series(age)
print(series)
```

```python
# Access Elements in Series based on "Index Location"
# "vari.loc[index_value]" --> Index Location
Age = pd.Series([20,24,26,19],index = ["a1","a2","a3","a4"])
print(Age.loc["a2"])
print(Age.loc["a4"])
```

```python
# Access Elements in Series based on "Integer Location"
# "vari.iloc[value]" --> Integer Location
Age = [18,19,20,21]
age = pd.Series(Age)
print(age.iloc[2])
print(age.iloc[0])
```

```python
# Filtering in Series - Condition Evaluation
# print(vari[vari > 'value'])
Age = [18,19,20,21]
age = pd.Series(Age)
print(age[age>18])
```

```python
# Pandas Series using "Dictionary as an Input"
dic_name = {'P1':10,'P2':20,'P3':30}
series = pd.Series(dic_name)
print(series)
```
